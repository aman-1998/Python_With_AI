### 1. Simple Inheritance

```python
class Computer:
    def __init__(self, brand, processor, ram):
        self.brand = brand
        self.processor = processor
        self.ram = ram

    def __str__(self):
        return f"Computer(brand={self.brand}, processor={self.processor}, ram={self.ram})"
    
class Laptop(Computer):
    def __init__(self, brand, processor, ram, battery_life):
        super().__init__(brand, processor, ram)
        self.battery_life = battery_life

    def __str__(self):
        return f"Laptop(brand={self.brand}, processor={self.processor}, ram={self.ram}, battery_life={self.battery_life})"
    

laptop1 = Laptop("Dell", "Intel i7", "16GB", "10 hours")
laptop2 = Laptop("HP", "Intel i5", "8GB", "8 hours")
laptop3 = Laptop("Apple", "M1", "16GB", "12 hours")

print(laptop1)
print(laptop2)
print(laptop3)
```

### 2. Multilevel Inheritance

```python
class Shape:
    def __init__(self, dimension):
        self.dimension = dimension
        
    def __str__(self):
        return f"Shape(dimension={self.dimension})"
    
class Sphere(Shape):
    def __init__(self, dimension):
        super().__init__(dimension)
        
    def __str__(self):
        return f"Sphere(dimension={self.dimension})"
    
class Cricket_Ball(Shape):
    def __init__(self, dimension, color):
        super().__init__(dimension)
        self.color = color
        
    def __str__(self):
        return f"Cricket_Ball(color={self.color}, dimension={self.dimension})"
    
sg_cricket_ball = Cricket_Ball(3, "Red")
print(sg_cricket_ball)
```

### 3. Hirearchical Inheritance

```python
class Animal:
    def __init__(self, common_name, species):
        self.common_name = common_name
        self.species = species

    def __str__(self):
        return f"Animal(common_name={self.common_name}, species={self.species})"
    
class Dog(Animal):
    def __init__(self, common_name, species, breed, color):
        super().__init__(common_name, species)
        self.breed = breed
        self.color = color
        
    def __str__(self):
        return f"Dog(common_name={self.common_name}, species={self.species}, breed={self.breed}, color={self.color})"
    
class Cat(Animal):
    def __init__(self, common_name, species, color):
        super().__init__(common_name, species)
        self.color = color
        
    def __str__(self):
        return f"Cat(common_name={self.common_name}, species={self.species}, color={self.color})"

    
dog1 = Dog("Pitbull", "Canis lupus familiaris", "Pitbull", "Grey-White")
dog2 = Dog("Dogo", "Canis lupus familiaris", "Dogo Argentino", "White")
dog3 = Dog("German Shepherd", "Canis lupus familiaris", "German Shepherd", "Black-Brown")

cat1 = Cat("Lion", "Panthera Leo", "Golden")
cat2 = Cat("Tiger", "Panthera Tigris", "Orange-Black")
cat3 = Cat("Leopard", "Panthera Pardus", "Spotted")
cat4 = Cat("Cheetah", "Acinonyx jubatus", "Spotted")
cat5 = Cat("Jaguar", "Panthera Onca", "Spotted")
cat6 = Cat("Snow Leopard", "Panthera Uncia", "Spotted")
cat7 = Cat("Cougar", "Puma concolor", "Tan")
cat8 = Cat("Street Cat", "Felis catus", "Grey")

# Printing all dog and cat objects
print(dog1)
print(dog2)
print(dog3)

# Printing all cat objects
print(cat1)
print(cat2)
print(cat3)
print(cat4)
print(cat5)
print(cat6)
print(cat7)
print(cat8)
```

### 4. Hybrid Inheritance (Combination of two or more inheritances)

```python
class Vehicle:
    def __init__(self, company, engine):
        self.company = company
        self.engine = engine

    def __str__(self):
        return f"Vehicle(company={self.company}, engine={self.engine})"

class Car(Vehicle):
    def __init__(self, company, engine, model):
        super().__init__(company, engine)
        self.model = model

    def __str__(self):
        return f"Car(company={self.company}, engine={self.engine}, model={self.model})"
    
class Bike(Vehicle):
    def __init__(self, company, engine, model, type):
        super().__init__(company, engine)
        self.model = model
        self.type = type

    def __str__(self):
        return f"Bike(company={self.company}, engine={self.engine}, model={self.model}, type={self.type})"
    
class Thar(Car):
    def __init__(self, company, engine, model, offroad_capable):
        super().__init__(company, engine, model)
        self.offroad_capable = offroad_capable

    def __str__(self):
        return f"Thar(company={self.company}, engine={self.engine}, model={self.model}, offroad_capable={self.offroad_capable})"
 
class Bullet(Bike):
    def __init__(self, company, engine, model, type, origin):
        super().__init__(company, engine, model, type)
        self.origin = origin

    def __str__(self):
        return f"Bullet(company={self.company}, engine={self.engine}, model={self.model}, type={self.type}, origin={self.origin})"
    

Thar_instance = Thar("Mahindra", "2000cc", "Thar Roxx", True)
Bullet_instance = Bullet("Royal Enfield", "350cc", "Bullet 350", "Cruiser", "India")

print("Thar Instance Details:")
print(Thar_instance.company)
print(Thar_instance.engine)
print(Thar_instance.model)
print(Thar_instance.offroad_capable)

print("\nBullet Instance Details:")
print(Bullet_instance.company)
print(Bullet_instance.engine)
print(Bullet_instance.model)
print(Bullet_instance.type)
print(Bullet_instance.origin)
    
```

### 5. Multiple Inheritance

##### Python uses MRO (Method Resolution Order) to decide which method to use.
##### <u>You can see the MRO with:</u>
##### print(child.mro())

child <br>
  ↓ <br>
Does child have skin_tone() ? <br>
  ↓ <br>
Mother <br>
  ↓ <br>
Does Mother have skin_tone() ? <br>
  ↓ <br>
Return "Fair skin tone" <br>

```python
class Mother:
    def __init__(self, name):
        self.name = name

    def cooking_skills(self):
        return "Excellent cooking skills"
    
    def skin_tone(self):
        return "Fair skin tone"
    
    def __str__(self):
        return f"Mother(name={self.name})"
    
class Father:
    def __init__(self, name):
        self.name = name

    def genetics(self):
        return "Strong genetics"
    
    def skin_tone(self):
        return "Black skin tone"
    
    def __str__(self):
        return f"Father(name={self.name})"
    
class Child(Mother, Father):
    def __init__(self, name):
        self.name = name
    
    # def skin_tone(self):
    #     return "Brown skin tone"
    
    def high_iq(self):
        return "High IQ"

    def __str__(self):
        return f"Child(name={self.name})"
    
child1 = Child("John")
print(child1)
print(child1.skin_tone())
print(child1.high_iq())
print(child1.genetics())
print(child1.cooking_skills())

# child
#   ↓
# Does child have skin_tone()? ❌
#   ↓
# Mother
#   ↓
# Does Mother have skin_tone()? ✅
#   ↓
# Return "Fair skin tone"
```

### Diamond Problem

```python
class A:
    def test1(self):
        print("This is test1 from class A")
    
class B(A):
    def test2(self):
        print("This is test2 from class B")
        
class C(A):
    def test3(self):
        print("This is test3 from class C")
        
class D(B, C):
    def test4(self):
        print("This is test4 from class D")
        
obj = D()
obj.test1()  # test1 will be resolved using MRO technique
obj.test2()
obj.test3()
obj.test4()

# D        
# ↓
# Does D has test1 method ? No
# ↓
# B
# ↓
# Does B has test1 method ? No
# ↓
# C
# ↓
# Does C has test1 method ? No
# ↓
# A
# ↓
# Does A has test1 method ? Yes
```

```python
class A:
    def test1(self):
        print("This is test1 from class A")
    
class B(A):
    def test2(self):
        print("This is test2 from class B")
        
    def test1(self):
        print("This is test1 from class B")
        
class C(A):
    def test3(self):
        print("This is test3 from class C")
        
class D(B, C):
    def test4(self):
        print("This is test4 from class D")
        
obj = D()
obj.test1()  # test1 will be resolved using MRO technique
obj.test2()
obj.test3()
obj.test4()

# D        
# ↓
# Does D has test1 method ? No
# ↓
# B
# ↓
# Does B has test1 method ? Yes
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def sound(self):
        return "Makes some sort of sound"

class Pet:
    def __init__(self, owner):
        self.owner = owner

class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def sound(self):
        return "Barks"
    
dog = Dog("Buddy", "Alice")

print(dog.name)
print(dog.owner)
print(dog.sound())
```
