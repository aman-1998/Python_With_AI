#### name   ===> Public
#### _name  ===> Protected
#### __name ===> Private

#### Always access the variable using getter and setters (Not mandatory but good practice)

```python
class Student:
    def __init__(self, name, age, subjects):
        self.__name = name
        self.__age = age
        self.__subjects = subjects

    def get_name(self):
        return self.__name
    
    def set_name(self, name):
        self.__name = name
    
    def get_age(self):
        return self.__age
    
    def set_age(self, age):
        self.__age = age
        
    def get_subjects(self):
        return self.__subjects
    
    def set_subjects(self, subjects):
        self.__subjects = subjects
        
    def __str__(self):
        return f"Name: {self.__name}, Age: {self.__age}, Subjects: {', '.join(self.__subjects)}"

   
student = Student("John Doe", 20, ["Math", "Science", "English", "Hindi"])
print(student.get_name())
print(student.get_age())
print(student.get_subjects())

student.set_name("Alice")
student.set_age(21)
student.set_subjects(["History", "Geography"])
print(student)

print(student.__name)  # This will raise an AttributeError because __name is private
```

```python
class Employee:
    def __init__(self, name, dept, salary):
        self._name = name
        self._dept = dept
        self._salary = salary

    def get_name(self):
        return self._name
    
    def set_name(self, name):
        self._name = name
        
    def get_dept(self):
        return self._dept
    
    def set_dept(self, dept):
        self._dept = dept
        
    def get_salary(self):
        return self._salary
    
    def set_salary(self, salary):
        self._salary = salary
    
    def __str__(self):
        return f"Employee(Name: {self._name}, Dept: {self._dept}, Salary: {self._salary})"
    

class Manager(Employee):
    def __init__(self, name, dept, salary, team_size):
        super().__init__(name, dept, salary)
        self.__team_size = team_size
        
    def get_team_size(self):
        return self.__team_size

    def set_team_size(self, team_size):
        self.__team_size = team_size
        
    def __str__(self):
        return f"Manager(Name: {self._name}, Dept: {self._dept}, Salary: {self._salary}, Team Size: {self.__team_size})"
        

manager = Manager("Alice", "IT", 120000, 14)
print(manager)
print("-----------------------------------------------------")
manager.set_team_size(20)
print(manager._name)
print(manager._dept)
print(manager._salary)
print(manager.get_team_size())
```
