### Polymorphism = Poly + Morphism = Many + Forms

```python
class Shape:
    def __init__(self, dimension):
        self.dimension = dimension
        
    def area(self):
        print("The area depends on the type of shape")
# ----------------------------------------------------------------------------        
class Square(Shape):
    def __init__(self, dimension, side_length):
        super().__init__(dimension)
        self.side_length = side_length
    
    # def area(self):
    #     print(f"The area of the square is {self.side_length ** 2}")
# -----------------------------------------------------------------------------
class Rectangle(Shape):
    def __init__(self, dimension, length, breadth):
        super().__init__(dimension)
        self.length = length
        self.breadth = breadth

    def area(self):
        print(f"The area of the rectangle is {self.length * self.breadth}")
# -------------------------------------------------------------------------------
class Circle(Shape):
    def __init__(self, dimension, radius):
        super().__init__(dimension)
        self.radius = radius
    
    def area(self):
        print(f"The area of the circle is {3.14 * self.radius ** 2}")
# --------------------------------------------------------------------------------
class Triangle(Shape):
    def __init__(self, dimension, base, height):
        super().__init__(dimension)
        self.base = base
        self.height = height

    def area(self):
        print(f"The area of the triangle is {0.5 * self.base * self.height}")
# -------------------------------------------------------------------------------
class Sphere(Shape):
    def __init__(self, dimension, radius):
        super().__init__(dimension)
        self.radius = radius

    def area(self):
        print(f"The surface area of the sphere is {4 * 3.14 * self.radius ** 2}")
# -------------------------------------------------------------------------------

circle = Circle("2D", 5)
circle.area()

square = Square("2D", 4)
square.area()

rectangle = Rectangle("2D", 5, 3)
rectangle.area()

triangle = Triangle("2D", 4, 3)
triangle.area()

sphere = Sphere("3D", 6)
sphere.area()
```

```python
# Function to calculate the area of any shape using polymorphism
def calculate_area(shape):
    shape.area()
    
circle = Circle("2D", 5)
calculate_area(circle)

square = Square("2D", 4)
calculate_area(square)

rectangle = Rectangle("2D", 5, 3)
calculate_area(rectangle)

triangle = Triangle("2D", 4, 3)
calculate_area(triangle)

sphere = Sphere("3D", 6)
calculate_area(sphere)
```

### Polymorphism with Abstract Base Class (ABC)

```python
from abc import ABC, abstractmethod

# Abstract base class for different types of sports
class Sports(ABC):
    @abstractmethod
    def no_of_players(self):
        pass
    
    @abstractmethod
    def duration(self):
        pass
    
    def format(self):
        pass
    
class Football(Sports):
    def no_of_players(self):
        print("A football team has 11 players")

    def duration(self):
        print("A football match lasts 90 minutes")
        
class Cricket(Sports):
    def no_of_players(self):
        print("A cricket team has 11 players")

    def duration(self):
        print("A cricket match lasts: \n\t1. 5 days in Test format\n\t2. 50 overs in One Day format\n\t3. 20 overs in T20 format")

    def format(self):
        print("Cricket has different formats: Test, One Day, and T20")
        
class Basketball(Sports):
    def no_of_players(self):
        print("A basketball team has 5 players")

    def duration(self):
        print("A basketball match lasts 48 minutes")
        

cricket = Cricket()
cricket.no_of_players()
cricket.duration()
cricket.format()

print("----------------------------------------------------------------------")

football = Football()
football.no_of_players()
football.duration()
football.format()

print("----------------------------------------------------------------------")

basketball = Basketball()
basketball.no_of_players()
basketball.duration()
basketball.format()
```
