#### We can achieve operator overloading by overriding below magic methods:
#### <strong>\_\_add__(self, other)</strong>     : Controls the behaviour of + operator
#### <strong>\_\_sub__(self, other)</strong>     : Controls the behaviour of - operator object
#### <strong>\_\_mul__(self, other)</strong>     : Controls the behaviour of * operator
#### <strong>\_\_truediv__(self, other)</strong> : Controls the behaviour of / operator
#### <strong>\_\_eq__(self, other)</strong>      : Controls the behaviour of == operator
#### <strong>\_\_lt__(self, other)</strong>      : Controls the behaviour of > operator
#### <strong>\_\_gt__(self, other)</strong>      : Controls the behaviour of > operator
#### <strong>\_\_ge__(self, other)</strong>      : Controls the behaviour of >= operator
#### <strong>\_\_le__(self, other)</strong>      : Controls the behaviour of <= operator

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented

    def __sub__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x - other.x, self.y - other.y)
        return NotImplemented

    def __mul__(self, scalar):
        if isinstance(scalar, (int, float)):
            return Vector(self.x * scalar, self.y * scalar)
        return NotImplemented

    def __truediv__(self, scalar):
        if isinstance(scalar, (int, float)):
            return Vector(self.x / scalar, self.y / scalar)
        return NotImplemented
    
    def __eq__(self, other):
        if isinstance(other, Vector):
            return self.x == other.x and self.y == other.y
        return NotImplemented
    
    def __lt__(self, other):
        if isinstance(other, Vector):
            return (self.x, self.y) < (other.x, other.y)
        return NotImplemented
    
    def __le__(self, other):
        if isinstance(other, Vector):
            return (self.x, self.y) <= (other.x, other.y)
        return NotImplemented

    def __gt__(self, other):
        if isinstance(other, Vector):
            return (self.x, self.y) > (other.x, other.y)
        return NotImplemented

    def __ge__(self, other):
        if isinstance(other, Vector):
            return (self.x, self.y) >= (other.x, other.y)
        return NotImplemented

    def __neg__(self):
        return Vector(-self.x, -self.y)

    def __repr__(self):
        return f"Vector(x={self.x}, y={self.y})"
```

```python
vector1 = Vector(3, 4)
vector2 = Vector(1, 2)

print("vector1 = " + str(vector1))
print("vector2 = " + str(vector2))

vector3 = vector1 + vector2
print("vector1 + vector2 = " + str(vector3))

vector4 = vector1 - vector2
print("vector1 - vector2 = " + str(vector4))

vector5 = vector1 * 2
print("vector1 * 2 = " + str(vector5))

vector6 = vector1 / 2
print("vector1 / 2 = " + str(vector6))

vector7 = -vector1
print("-vector1 = " + str(vector7))

vector8 = vector1 == vector2
print("vector1 == vector2 = " + str(vector8))

vector9 = vector1 < vector2
print("vector1 < vector2 = " + str(vector9))

vector10 = vector1 <= vector2
print("vector1 <= vector2 = " + str(vector10))

vector11 = vector1 > vector2
print("vector1 > vector2 = " + str(vector11))

vector12 = vector1 >= vector2
print("vector1 >= vector2 = " + str(vector12))
```
