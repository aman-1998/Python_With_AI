#### Magic methods are predefined methods in python that you can override to change the behaviour of your objects. Some of the common magic methods include :

#### <strong>\_\_init__</strong>    : Initializes new instance of a class
#### <strong>\_\_str__</strong>     : Returns a string representation of an object
#### <strong>\_\_repr__</strong>    : Returns an official string representation of an object
#### <strong>\_\_eq__</strong>      : Controls what happens when you use the == operator
#### <strong>\_\_hash__</strong>    : It determines the hash value of an object, which is used by data structures such as set, dict, etc
#### <strong>\_\_len__</strong>     : Returns the length of an object
#### <strong>\_\_del__</strong>     : This method is called when an object is about to be destroyed/garbage-collected.

```python
class Student:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def __str__(self):
        return f"Student[Name: {self.__name}, Age: {self.__age}]"

    def __repr__(self):
        return f"Student(name={self.__name!r}, age={self.__age!r})"
    
    def __eq__(self, other):
        if isinstance(other, Student):
            return self.__name == other.__name and self.__age == other.__age
        return False
    
    def __hash__(self):
        return hash((self.__name, self.__age))
    
    def __len__(self):
        return len(self.__name)
    
    def __del__(self):
        print(f"Student object with name {self.__name} is being deleted")
    
    
student1 = Student("John", 20)
print(student1)
print(repr(student1))

student2 = Student("John", 20)
print(student1 == student2)
print(hash(student1))
print(hash(student2))

print(len(student1))
```

<strong>del student1</strong> does not necessarily destroy the object instantly.

It immediately removes the variable/reference named student (reference count decreases). What happens to the object afterward depends on whether other references to that object still exist. In CPython, if reference count becomes zero then the object is immediately deleted or else it remains in the memory.

```python
import sys

del student1

print("Reference count of Student = ", sys.getrefcount(Student))
```
