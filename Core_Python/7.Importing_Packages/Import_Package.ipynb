```python
import random

print(random.randint(1, 10)) # Generate a random integer between 1 and 10
print(random.random()) # Generate a random float between 0 and 1
print(random.choice([1, 2, 3, 4, 5])) # Randomly select an element from the list
lst = [1, 2, 3, 4, 5]
random.shuffle(lst) # Shuffle the list in place
print(lst)
```

```python
import re

pattern = r'\d+'
text = "There are 123 numbers in this 456 text."
matches = re.findall(pattern, text)
print(matches) # Output: ['123', '456']

pattern2 = r'\b\w{4}\b'
matches2 = re.findall(pattern2, text)
print(matches2) # Output: ['This', 'text']

match = re.search(r'\b\w{3}\b', text)
print(match) # Output: <re.Match object; span=(4, 7), match='are'>
print(match.group()) # Output: 'are'

print(bool(match)) # Output: True
```
