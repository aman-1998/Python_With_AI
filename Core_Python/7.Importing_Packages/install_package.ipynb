```python
import numpy as np

print(np.array([1, 2, 3, 4])) # Create a NumPy array from a Python list
print(np.zeros(5)) # Create an array of zeros
print(np.ones(5)) # Create an array of ones
print(np.arange(10)) # Create an array with a range of values
print(np.linspace(0, 1, 5)) # Create an array with linearly spaced values
print(np.random.rand(5)) # Create an array of random values between 0 and 1
print(np.random.randint(0, 10, 5)) # Create an array of random integers
print(np.random.randn(0, 5)) # Create an array of random values from the standard normal distribution
print(np.eye(5)) # Create an identity matrix
print(np.diag([1, 2, 3, 4, 5])) # Create a diagonal matrix
print(np.triu(np.ones((5, 5)))) # Create an upper triangular matrix
print(np.tril(np.ones((5, 5)))) # Create a lower triangular matrix
print(np.full((5, 5), 7)) # Create an array filled with a specific value
print(np.identity(5)) # Create an identity matrix
print(np.random.seed(42)) # Set the random seed for reproducibility
```
