### Operators in Python:
Arithmetic: &nbsp;       + &nbsp; - &nbsp; * &nbsp; / &nbsp; // &nbsp; % &nbsp; **
<br>Comparison: &nbsp;       == &nbsp; != &nbsp; > &nbsp; < &nbsp; >= &nbsp; <=
<br>Assignment: &nbsp;       = &nbsp; += &nbsp; -= &nbsp; *= &nbsp; /= &nbsp; //= &nbsp; %= &nbsp; **=
<br>Logical: &nbsp;         and &nbsp; or &nbsp; not
<br>Identity: &nbsp;        is &nbsp; is not
<br>Membership: &nbsp;       in &nbsp; not in
<br>Bitwise: &nbsp;          & &nbsp; | &nbsp; ^ &nbsp; ~ &nbsp; << &nbsp; >>

```python
# Arithmetic operators

print(4+7) # Addition
print(5-3) # subtraction
print(7*3) # multiplication
print(5/2) # division
print(5//2) # floor division
print(5%2) # modulo or remainder
print(4**3) # power
```

```python
# Comparison operators

a = 3
b = 3
print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
# Assignment operators

a = 4
print(a)

a += 3
print(a)

b = 10
b -= 5
print(b)

b = 10
b *= 5
print(b)

b = 10
b /= 5
print(b)

b = 10
b //= 5
print(b)

b = 10
b %= 4
print(b)

b = 2
b **= 4
print(b)
```

```python
# Logical operator

a = True
b = False

print(a or b)
print(a and b)
print(not(a))
```

```python
# Identity operator

a = [1,2,3]
b = [1,2,3]

print(a == b)
print(a is b)
print(a is not b)
```

```python
# Membership operator

a = [1,2,3,4]

print(2 in a)
print(5 not in a)
```

```python
# Bitwise operator (&  |  ~  ^  <<  >>)

print(2 & 1)
print(2 | 1)
print(~1)
print(2 ^ 1)
print(2 << 1)
print(2 >> 1)
```
