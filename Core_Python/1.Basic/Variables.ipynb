### Variables must contain alphanumeric or underscore( \_ ). It must start with either alphabet or underscore( \_ )

```python
name_123 = 'Aman Mishra'
print(name_123)

_title = 'Kumar'
print(_title)
```

### Input and type conversion

```python
num = input('Enter any number')
print(num)

print(type(num))
```

```python
num = int(input('Enter any number'))
print(num)

print(type(num))
```

```python
num = '1236'
print(int(num))
```

```python
num = '956A'
print(int(num))
```

```python
name = 'Aman Mishra'
print(int(name))
```

```python
num = 27203
print(str(num))

print(type(num))

print(type(str(num)))
```
