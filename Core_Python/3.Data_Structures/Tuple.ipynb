### Tuple
Note: Tuples are immutable in nature<br>
Note: Slicing and iteration concepts are same as List

```python
tpl = (7,4,2,3,5,1)

for i in tpl:
    print(str(i) + str(" "), end="")
```

```python
lst = [1,4,6,2,3,7]
tpl = tuple(lst)

n = len(tpl)

for i in range(n):
    print(str(tpl[i]) + str(" "), end="")
```

```python
tpl = (4, 'Hello', 'Water', 5, 3.7, 'Drop')

for i,_ in enumerate(tpl):
    print(str(tpl[i]) + str(" "), end="")
```

```python
tpl = (4, 'Hello', 'Water', 5, 3.7, 'Drop')

tpl[1] = 2
```

```python
# Packing of Tuple

tpl = 5, 'Nice', 8, 3.5, 'Good'

for i in tpl:
    print(str(i) + str(" "), end="")
```

```python
# Unpacking of Tuple

a,b,c,d,e = tpl

print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
# Unpacking of Tuple

a,*b,d,e = tpl

print(a)
print(b)
print(d)
print(e)
```
