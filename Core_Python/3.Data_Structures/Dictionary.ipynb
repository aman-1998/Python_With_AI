### Shallow copy

What gets shared ? <br>
| Object type | Shared after `.copy()`? |
| --- | --- |
| `int` | Yes, but it's immutable |
| `float` | Yes, but it's immutable |
| `bool` | Yes, but it's immutable |
| `str` | Yes, but it's immutable |
| `tuple` | Yes, but it's immutable |
| `list` | Yes |
| `dict` | Yes |
| `set` | Yes |

```python
dic = {'name':'Aman', 'title':'Mishra', 'age':28, 'department':'CSE'}

dic_copy = dic # Here, both dic and dic_copy are referring to the same dictionary object

dic_copy['age'] = 30

print(dic)
print(dic_copy)
```

```python
dic = {'name':'Aman', 'title':'Mishra', 'age':28, 'department':'CSE'}

dic_cpy = dic.copy() # Shalow copy : dic_cpy and dic are referring to different dictianary objects

# Here, we are not changing the integer(28) object, instead we are making the age key refer to 
# different integer(30).
dic_cpy['age'] = 30

print(dic)
print(dic_cpy)
```

```python
dic = {
    'name': 'Aman Mishra',
    'marks': [80, 90, 76, 84],
    'subjects': ['Physics', 'Math', 'Chemistry', 'English'],
    'age': 28
}

dic_cpy = dic.copy() # Shallow copy

dic_cpy['marks'][0] = 85
dic_cpy['age'] = 30

print(dic)
print(dic_cpy)
```

```python
import copy

dic = {
    'name': 'Aman Mishra',
    'marks': [80, 90, 76, 84],
    'subjects': ['Physics', 'Math', 'Chemistry', 'English'],
    'age': 28
}

dic_cpy = copy.deepcopy(dic) # Deep copy

dic_cpy['marks'][0] = 85
dic_cpy['age'] = 30

print(dic)
print(dic_cpy)
```

### Dictionary Comprehension
<u>Syntax:</u>
<br>{key_expression: value_expression for item in iterable}
<br>{key_expression: value_expression for item in iterable condition}
<br>{key_expression: value_expression for item1 in iterable1 for item2 in iterable2}
<br>{key_expression: value_expression for item1 in iterable1 if condition1 for item2 in iterable2}
<br>{key_expression: value_expression for item1 in iterable1 if condition1 for item2 in iterable2 if condition2}

```python
dic = {x: x**2 for x in range(5)}

print(dic)

# dic = {}
# for x in range(5):
#     dic[x] = x**2

# print(dic)
```

```python
dic = {str('key')+str(x): x+2 for x in range(10) if x%2 == 0}

print(dic)

# dic = {}
# for x in range(10):
#     if x%2 == 0:
#         dic[str('key')+str(x)] = x+2

# print(dic)
```

```python
dic = {str('key')+str(x+i): x**3 for x in range(5) for i in range(3)}

print(dic)

# dic = {}
# for x in range(5):
#     for i in range(3):
#         dic[str('key')+str(x+i)] = x**3

# print(dic)
```

```python
dic = {str('key')+str(x+i): x**3 for x in range(5) if x%2 == 0 for i in range(3) if i%2 != 0}

print(dic)

# dic = {}
# for x in range(5):
#     if x%2 == 0:
#         for i in range(3):
#             if i%2 != 0 :
#                 dic[str('key')+str(x+i)] = x**3

# print(dic)
```

```python
# Find frequency of every elements

lst = [1,5,3,1,6,5,5,8,9,8,1]
dic = {} # key=element, value=count

for i in lst:
    freq = dic.get(i)
    if freq == None:
        dic[i] = 1
    else:
        dic[i] = freq+1
        
print(dic)
        
```
