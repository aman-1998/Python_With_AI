### Slicing 

```python
fruits = ["Apple", "Mango", "Guava", "Kiwi", "Coconut", "Avocado", "Orange", "Banana", "Grapes"]

print(fruits[:3]) # 0 to 2
print(fruits[2:5]) # 2 to 4
print(fruits[2:]) # 2 to last_index
print(fruits[:]) # 0  to last_index
print(fruits[1:8:2]) # 1 to 7 with step 2
print(fruits[8:1]) # 8 to 0
print(fruits[8:2:-1]) # 8 to 3 with step -1
print(fruits[8:0:-2]) # 8 to 1 with step -2
print(fruits[1::2]) # 1 to last_index with step 2
print(fruits[::3]) # 0 to last_index with step 3
print(fruits[::-1]) # last_index to 0 with step -1
print(fruits[::-3]) # last_index to 0 with step -3
print(fruits[:6:2]) # 0 to 5 with step 2
print(fruits[:3:-1]) # last_index to 4 with step -1

```

```python
numbers = [5, 4, 3, 2, 1]

for i,_ in enumerate(numbers): # _ is used to ignore the returned value
    print(numbers[i])
```

```python
vegetables = ['Radish', 'Potato', 'Pumpkin', 'Spinach', 'Cauliflower', 'Cabbage']

for i,veg in enumerate(vegetables): # _ is used to ignore returned value
    print(str(i) + ', ' + veg)
```

### List comprehension
<u>Syntax:</u><br>
<br>1. [expression for item in iterable]
<br>2. [expression for item in iterable condition]
<br>3. [expression for item1 in iterable1 for item2 in iterable2]
<br>4. [expression for item1 in iterable1 condition1 for item2 in iterable2]
<br>4. [expression for item1 in iterable1 condition1 for item2 in iterable2 condition2]

```python
lst = [i**2 for i in range(5)]

for i in lst:
    print(i)
    

#lst = []
#for i in range(5):
#    print(i**2)
```

```python
lst = [i for i in range(11) if i%2 == 0]

for i in lst:
    print(i)
    

#lst = []
#for i in range(5):
#    if i%2 == 0:
#        print(i)
```
