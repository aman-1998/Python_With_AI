### Reading from a file

```python
with open("test/example.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
with open("test/example.txt", "r") as file:
    for line in file:
        print(line, end='') # Print the line as is, without adding extra newline characters
        # print(line.strip("\n")) # Remove only the newline character at the end of the line
        # print(line.strip()) # Remove all leading and trailing whitespace characters
```

```python
with open("test/example2.txt", "w") as file:
    file.write("Hello !!\n")
    file.write("\n")
    file.write("I am Abhishek Bhardwaj.\n")
    file.write("I work as Platform Engineer in Google.\n")
    file.write("I completed my B.Tech in 2016.\n")
```

```python
lines = ["First line\n", "Second line\n", "Third line\n"]

with open("test/example3.txt", "w") as file:
    file.writelines(lines)
```

```python
lines = b"First line\nSecond line\nThird line\n"

with open("test/example4.txt", "wb") as file:
    file.write(lines)
```

```python
# Transfer from source file to destination file
with open("test/example2.txt", "r") as source_file:
    lines = source_file.readlines()

with open("test/example5.txt", "w") as destination_file:
    destination_file.writelines(lines)
```

```python
# Count lines, words and characters in a file
def count(file_path):
    with open(file_path, "r") as file:
        lines = file.readlines()

    line_count = len(lines)
    word_count = sum(len(line.split()) for line in lines)
    char_count = sum(len(line) for line in lines)
    
    return line_count, word_count, char_count

line_count, word_count, char_count = count("test/example2.txt")

print(f"Lines: {line_count}, Words: {word_count}, Characters: {char_count}")
```

```python
# Open the file in write and read mode, write content, then read it from the beginning
with open("test/example6.txt", "w+") as file:
    file.write("I am Alice.\n")
    file.write("I study in Wonderland.\n")
    
    file.seek(0) # Move the file pointer to the beginning of the file
    
    content = file.read()
    print(content)
```
