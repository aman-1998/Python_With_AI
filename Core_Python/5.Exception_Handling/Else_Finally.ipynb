### Use of Else and Finally block

```python
try:
    x = 12/int(input("Enter the denomenator: "))
except ZeroDivisionError as ex:
    print(f"Cannot divide by zero: {ex}")
except ValueError as ex:
    print(f"Invalid input. Please enter an integer: {ex}")
except Exception as ex:
    print(f"An unexpected error occurred: {ex}")
else:
    print(f"The result is: {x}") # else block executes only if no exception occurs
```

```python
try:
    x = 12/int(input("Enter the denomenator: "))
except ZeroDivisionError as ex:
    print(f"Cannot divide by zero: {ex}")
except ValueError as ex:
    print(f"Invalid input. Please enter an integer: {ex}")
except Exception as ex:
    print(f"An unexpected error occurred: {ex}")
else:
    print(f"The result is: {x}") # else block executes only if no exception occurs
finally:
    print("Execution completed.") # finally block executes regardless of exceptions
```
