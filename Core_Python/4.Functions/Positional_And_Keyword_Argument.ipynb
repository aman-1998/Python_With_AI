### Positional Argument

```python
def test1(*args):
    for arg in args:
        print(arg)
        
test1(1, 2, 3, "Ritesh", [4, 5, 6], {"title": "Mishra"}, None)
```

```python
def test2(**kargs):
    for karg in kargs.items():
        print(karg)

test2(a=1, b=2, c=3, name="Ritesh", data=[4, 5, 6], info={"title": "Mishra"}, empty=None)
```

```python
def test3(**kargs):
    for key, value in kargs.items():
        print(f"{key} : {value}")
        print(str(type(key)) + " : " + str(type(value)))
        print("-----------------------------------")
        
test3(a=1, b=2, c=3, name="Ritesh", data=[4, 5, 6], info={"title": "Mishra"}, empty=None)
```
