### <u>map()</u>
Map Function is used for data transformation

```python
listOfInteger = [1, 2, 3, 4, 5]

squares = map(lambda x: x**2, listOfInteger)

print(list(squares))
```

```python
def convertToWord(num):
    words = ["one", "two", "three", "four", "five", "six", "seven", "eight", "nine", "ten"]
    return words[num - 1]

listOfInteger = [4, 10, 7, 1, 8, 2, 9, 8, 5, 3, 4, 8, 3, 6, 10, 6]

listOfWords = list(map(lambda x: convertToWord(x), listOfInteger))

print(listOfWords)
```

```python
numbers1 = [1, 2, 3, 4, 5]
numbers2 = [6, 7, 8, 9, 10]

sums = list(map(lambda x, y: x + y, numbers1, numbers2))
print(sums)
```

```python
numbers1 = [1, 2, 3, 4, 5]
numbers2 = [6, 7, 8, 9, 10, 11, 12, 13] # Note: numbers1 and numbers2 have different lengths

sums = list(map(lambda x, y: x + y, numbers1, numbers2)) # Only sums elements up to the length of the shorter list
print(sums)
```

```python
listOfNumbersStr = ["1", "2", "3", "4", "5"]

print(listOfNumbersStr)

listOfNumbersInt = list(map(int, listOfNumbersStr))

print(listOfNumbersInt)
```

```python
names = ["Alice", "Bob", "Charlie", "David", "Eve"]

print(names)

nameInUpper = list(map(str.upper, names))

print(nameInUpper)
```

```python
def get_names(person):
    return person["name"]

people = [
    {"name": "Alice", "age": 30, "city": "New York"},
    {"name": "Bob", "age": 25, "city": "Los Angeles"},
    {"name": "Charlie", "age": 35, "city": "Chicago"}
]

names = list(map(get_names, people))

print(names)
```
