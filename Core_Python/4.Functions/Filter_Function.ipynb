### filter()

Filter function is used to filter from iterables

```python
def isEven(number):
    return number % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]

evenNumbers = list(filter(isEven, numbers))

print(evenNumbers)
```

```python
numbers = [1, 2, 3, 4, 5, 6]

oddNumbers = list(filter(lambda x : x % 2 != 0, numbers))

print(oddNumbers)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

greaterThanFive = list(filter(lambda x : x > 5, numbers))

print(greaterThanFive)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

evenGreaterThanFive = list(filter(lambda x : x % 2 == 0 and x > 5, numbers))

print(evenGreaterThanFive)
```

```python
people = [
    {"name": "Alice", "age": 10, "city": "New York"},
    {"name": "Bob", "age": 13, "city": "Los Angeles"},
    {"name": "Charlie", "age": 35, "city": "Chicago"},
    {"name": "David", "age": 18, "city": "Miami"},
    {"name": "Eve", "age": 30, "city": "San Francisco"},
    {"name": "Frank", "age": 17, "city": "Boston"}
]

adults = list(filter(lambda person : person["age"] >= 18, people))

print(adults)
```
