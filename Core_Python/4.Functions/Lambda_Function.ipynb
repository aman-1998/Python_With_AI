### Lambda Function

```python
def add(a, b):
    return a + b

sum = add(4, 7)

print("result", sum)

# Using a lambda function to achieve the same result

add = lambda a, b: a + b

sum = add(4, 7)

print("result:", sum)
```

```python
test = lambda x: "Even" if x%2 == 0 else "Odd"

print(test(4))
print(test(7))
```

```python
isEven = lambda num: num%2 == 0

print(isEven(4))
print(isEven(7))
```
