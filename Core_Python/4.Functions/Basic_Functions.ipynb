## Function Syntax

```python
def sum(a, b):
    print(a+b)
    
res = sum(3, 5)

print(res)
```

```python
def test(name):
    print(f"Hello, {name}!")
    
test("Alice")
test("Bob")
test("Charlie")
```

```python
def calculator1(a, b, operation):
    switcher = {
        "add": a + b,
        "subtract": a - b,
        "multiply": a * b,
        "divide": a / b if b != 0 else None
    }
    return switcher.get(operation, None)

print(calculator1(10, 5, "add"))
print(calculator1(10, 5, "subtract"))
print(calculator1(10, 5, "multiply"))
print(calculator1(10, 5, "divide"))
```

```python
def calculator2(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        # return a / b if b != 0 else None
        if(b != 0):
            return a / b
        else:
            return None
    else:
        return None
    
res = calculator2(10, 5, "+")
print(res)
res = calculator2(10, 5, "-")
print(res)
res = calculator2(10, 5, "*")
print(res)
res = calculator2(10, 5, "/")
print(res)
```
